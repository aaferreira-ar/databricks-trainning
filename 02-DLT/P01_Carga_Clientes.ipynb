```sql
-- In Delta Live Tables, the catalog and schema are set at the pipeline level in the pipeline settings,
-- not directly in the Python code. However, you can specify the target schema in Python using the @dlt.table decorator.

-- Example: Specify the target schema in Python for a DLT table
import dlt

@dlt.table(
  comment="Example table in specific catalog and schema",
  table_properties={
    "delta.catalog": "your_catalog_name",
    "delta.schema": "your_schema_name"
  }
)
def example_table():
    return spark.sql("SELECT 1 AS example_column")

-- Note: Replace 'your_catalog_name' and 'your_schema_name' with your actual catalog and schema names.

-- To set the catalog for the entire pipeline, configure it in the Delta Live Tables pipeline settings UI or JSON configuration.
```

```sql

@dlt.table(
  comment="Example table in specific catalog and schema",
  table_properties={
    "delta.catalog": settings['lab.catalog'],
    "delta.schema": settings['lab.schema']
  }
)
def customers():
    """
    Reads the raw sample orders JSON data as a streaming source.
    """
    path = "/databricks-datasets/retail-org/sales_orders/"

    schema = """
        customer_id INTEGER,
        customer_name STRING,
        email STRING,
        address STRING,
        city STRING,
        state STRING,
        zip STRING,
        phone STRING
    """

    return (
        spark.readStream.schema(schema)
            .format("csv")
            .option("header", "true")
            .load(path)
    )
```
