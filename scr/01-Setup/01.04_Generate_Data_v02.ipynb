Rdatasets é uma coleção de 2337 conjuntos de dados que foram originalmente distribuídos juntamente com o ambiente de software estatístico "R" e alguns dos seus pacotes complementares. O objetivo é tornar estes dados mais amplamente acessíveis para o ensino e desenvolvimento de software estatístico.
