```python
import os
import shutil
import pandas as pd
import random

from faker import Faker
```

### Gerar dataset: Clientes (customers)
- Serão gerados 100.000 clientes 
- OS clientes serão garvados em 10 arquivos csv no volumes files

```python
fake = Faker()
Faker.seed(0)
customers_ids = 100000

data = {
    "customer_id": list(range(1, customers_ids + 1)),
    "customer_name": [fake.name() for _ in range(customers_ids)],
    "email": [fake.email() for _ in range(customers_ids)],
    "address": [fake.street_address() for _ in range(customers_ids)],
    "city": [fake.city() for _ in range(customers_ids)],
    "state": [fake.state() for _ in range(customers_ids)],
    "zip": [fake.zipcode() for _ in range(customers_ids)],
    "phone": [fake.phone_number() for _ in range(customers_ids)]
}

customers_df = pd.DataFrame(data)
# display(customers_df)
```

```python
output_path = f"{settings['lab.path.files']}/generated/customers"
os.makedirs(output_path, exist_ok=True)

file_name = f"{output_path}/customers.csv"
customers_df.to_csv(file_name, index=False)
```

### Gerar dataset Vendas
- Para Cada cliente gerar entre 100 e 1000 vendas
- Gravar os arquivos por data

```python

output_path = f"{settings['lab.path.files']}/generated/sales"
os.makedirs(output_path, exist_ok=True)

start_date = '2025-01-01'

for i in range(60):
    sale_date = pd.to_datetime(start_date) + pd.DateOffset(days=i)

    qtd_rows = random.randint(1000, 100000)

    data = {
        "customer_id": [random.randint(1, customers_ids) for _ in range(qtd_rows)],
        "sale_date": [sale_date for _ in range(qtd_rows)],
        "product": [fake.word() for _ in range(qtd_rows)],
        "quantity": [random.randint(1, 30) for _ in range(qtd_rows)],
        "value": [round(random.uniform(10.0, 1000.0), 2) for _ in range(qtd_rows)]
    }

    sales_df = pd.DataFrame(data)

    file_name = f"{output_path}/sales_{sale_date.strftime('%Y%m%d')}.csv"
    sales_df.to_csv(file_name, index=False)
```
