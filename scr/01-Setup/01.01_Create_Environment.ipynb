```python
import os
```

```python
# Criar Schema
# Ref.: https://docs.databricks.com/aws/pt/schemas/create-schema

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {settings['lab.catalog']}.{settings['lab.schema']}")
```

```python
# Create volume files
# Ref.: https://docs.databricks.com/aws/pt/volumes/

spark.sql(f"""
CREATE VOLUME IF NOT EXISTS {settings['lab.catalog']}.{settings['lab.schema']}.files
COMMENT 'Volume for storing files'
""")
```
