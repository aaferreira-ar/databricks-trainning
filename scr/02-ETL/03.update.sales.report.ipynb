```python
query = f"""
  CREATE OR REPLACE TABLE {settings['lab.catalog']}.{settings['lab.schema']}.sales_resume AS
  SELECT
      c.city,
      sum(s.value) as value
  FROM {settings['lab.catalog']}.{settings['lab.schema']}.sales as s
  LEFT JOIN {settings['lab.catalog']}.{settings['lab.schema']}.customers as c
  ON s.customer_id = c.customer_id
  group by all
"""

print(query)
```

```python
spark.sql(query)
```
