```python
import os
import shutil
import glob
```

```python
def move_archiving(file_list):
    files_path = os.path.dirname(file_list[0])
    os.makedirs(files_path.replace("/landing/", "/archiving/"), exist_ok=True)

    for file_name in file_list:
        shutil.move(file_name, file_name.replace("/landing/", "/archiving/"))
```

```python
def read_files(file_list, table_name):

    df = (spark.read.format("csv")
               .option("header", "true")
               .option("inferSchema", "true")
               .load(file_list))
    
    df.write.mode("append").saveAsTable(table_name)
```

```python
# Obter lista de arquivos
files_path=f"{settings['lab.path.files']}/etl/landing/sales/sales*.csv"

file_list = glob.glob(files_path)
file_list
```

```python
# Ler arquivos
# Carregar para tabela Delta
# Mover arquivos para archivong

table_name = f"{settings['lab.catalog']}.{settings['lab.schema']}.sales"

if file_list != []:
    read_files(file_list, table_name)
    move_archiving(file_list)
else:
    print("Nenhum arquivo para carregar.")
```
