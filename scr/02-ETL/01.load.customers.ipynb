```python
import shutil
import glob
```

```python
# Obter lista de arquivos
file_list = glob.glob(f"{settings['lab.path.files']}/etl/landing/customers/customers*.csv")
file_list
```

```python

table_name = f"{settings['lab.catalog']}.{settings['lab.schema']}.customers"

if file_list != []:
    df = (spark.read.format("csv")
               .option("header", "true")
               .option("inferSchema", "true")
               .load(file_list))
    df.write.mode("append").saveAsTable(table_name)
else:
    print("Nenhum arquivo para carregar.")
```

```python
# Mover arquivos para archiving
for file_name in file_list:
    dbutils.fs.mv(file_name, file_name.replace("/landing/", "/archiving/"))
```
